```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
import seaborn as sns
```

```python
np.random.seed(613)
x = np.arange(50)
y = np.random.randint(0, 100, 50)
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y)
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.bar(x,y)
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.hist(y)  #creating a histogram plot of y frequency
ax.hist(y, edgecolor='black')
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(y)
ax.set_title('Total growth over time')
ax.set_ylabel('Total growth')
ax.set_xlabel('Years since start')
fig.tight_layout()
```

```python
font1 = {'family':'sans-serif','color':'blue','size':20}
font2 = {'family':'monospace','color':'green','size':14}
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)
ax.set_title('Total growth over time', fontdict = font1)
ax.set_ylabel('Total growth', fontdict = font2)
ax.set_xlabel('Years since start', fontdict = font2)
fig.tight_layout
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x, 
           y, 
           marker='*', # NEW ADDITIONS 
           color = "indigo") # NEW ADDITIONS
fig.show()
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,
        y,
        marker='*',
        color = 'indigo',
        linestyle='--',
        linewidth=2)
fig.show()
```

```python
fig, ax = plt.subplots(figsize=(5, 3))  #creating a figure and axis object with specified size

ax.plot(x,
        y,
        marker='*',
        color = '#7425b9',
        linestyle='--',
        linewidth=2)
fig.show()
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,
        y,
        marker='*',
        markersize = 12,
        color = '#7425b9',
        linestyle='--',
        linewidth= 2,
        markeredgecolor = '#fa9359', # NEW ADDITIONS
        markerfacecolor = '#000000' ) # NEW ADDITIONS
ax.grid(axis = 'y', color = "blue", linewidth = 2, linestyle = '-.')
fig.show()
```
